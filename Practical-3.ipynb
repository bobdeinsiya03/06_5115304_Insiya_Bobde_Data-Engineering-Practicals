```python
# ============================================================
# PRACTICAL 3: MongoDB Operations
# ============================================================

# Install PyMongo
!pip install -q pymongo mongomock

# Import libraries
import mongomock
from pprint import pprint

print("PyMongo environment ready!")

# ============================================================
# Create a MongoDB-like local database for Colab
# ============================================================

client = mongomock.MongoClient()

db = client["DE_Practical_03"]

print("Connected to MongoDB successfully!")

# ============================================================
# STEP 1: Insert Single Document
# ============================================================

print("\n--- Step 1: Insert Single Document ---")

items = db["items"]

result = items.insert_one({
    "name": "laptop",
    "price": 999
})

print("Document inserted successfully!")
print("Inserted ID:", result.inserted_id)

# ============================================================
# STEP 2: View Inserted Document
# ============================================================

print("\n--- Step 2: View Inserted Document ---")

for document in items.find():
    pprint(document)

# ============================================================
# STEP 3: Insert Multiple Documents
# ============================================================

print("\n--- Step 3: Insert Multiple Documents ---")

products = db["products"]

data = [
    {
        "name": "phone",
        "price": 500,
        "stock": 10
    },
    {
        "name": "tablet",
        "price": 300,
        "stock": 5
    },
    {
        "name": "watch",
        "price": 150,
        "stock": 0
    }
]

result = products.insert_many(data)

print("Multiple documents inserted successfully!")

print("Inserted IDs:")
for inserted_id in result.inserted_ids:
    print(inserted_id)

# ============================================================
# STEP 4: View Products
# ============================================================

print("\n--- Step 4: View Products ---")

for document in products.find():
    pprint(document)

# ============================================================
# COMPLETION
# ============================================================

print("\n==========================================")
print("PRACTICAL 3 COMPLETED SUCCESSFULLY!")
print("==========================================")
```

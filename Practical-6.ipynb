```python
#Que: Setting up Apache Airflow and Creating a DAG1. ObjectiveConfigure a Directed Acyclic Graph (DAG) in Apache Airflow to run daily, executing sequential dependency tasks: a start notifier, an extraction script, and a pipeline success logger.2. DAG ImplementationPlace this code in your Airflow DAGs directory (usually ~/airflow/dags/data_pipeline_dag.py):pythonfrom datetime import datetime, timedelta
from datetime import datetime, timedelta

# Install required packages, including explicit installation of rfc3987
!pip install --upgrade pip
!pip install rfc3987 apache-airflow --upgrade

from airflow import DAG
from airflow.operators.bash import BashOperator
from airflow.operators.empty import EmptyOperator


# Default arguments
default_args = {
    "owner": "data_engineering_lab",
    "depends_on_past": False,
    "start_date": datetime(2026, 1, 1),
    "email_on_failure": False,
    "email_on_retry": False,
    "retries": 1,
    "retry_delay": timedelta(minutes=5),
}


# Define DAG
with DAG(
    dag_id="university_etl_orchestration",
    default_args=default_args,
    description="Daily ETL pipeline orchestration",
    schedule=timedelta(days=1),
    catchup=False,
) as dag:

    # Task 1: Start pipeline
    start_pipeline = EmptyOperator(
        task_id="start_pipeline"
    )

    # Task 2: Run extraction script
    execute_extraction = BashOperator(
        task_id="run_extraction_script",
        bash_command="python3 /home/$USER/airflow/dags/data_extraction.py",
    )

    # Task 3: Log successful completion
    pipeline_complete = BashOperator(
        task_id="log_pipeline_success",
        bash_command='echo "ETL Execution completed successfully at $(date)"',
    )

    # Task dependencies
    start_pipeline >> execute_extraction >> pipeline_complete
```

```python
from datetime import datetime, timedelta

from airflow import DAG
from airflow.operators.bash import BashOperator
from airflow.operators.empty import EmptyOperator


default_args = {
    "owner": "data_engineering_lab",
    "depends_on_past": False,
    "start_date": datetime(2026, 1, 1),
    "email_on_failure": False,
    "email_on_retry": False,
    "retries": 1,
    "retry_delay": timedelta(minutes=5),
}


with DAG(
    dag_id="university_etl_orchestration",
    default_args=default_args,
    description="Daily ETL pipeline orchestration",
    schedule=timedelta(days=1),
    catchup=False,
) as dag:

    start_pipeline = EmptyOperator(
        task_id="start_pipeline"
    )

    execute_extraction = BashOperator(
        task_id="run_extraction_script",
        bash_command="python3 /home/$USER/airflow/dags/data_extraction.py",
    )

    pipeline_complete = BashOperator(
        task_id="log_pipeline_success",
        bash_command='echo "ETL Execution completed successfully at $(date)"',
    )

    start_pipeline >> execute_extraction >> pipeline_complete
```

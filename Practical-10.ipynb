```python
import pandas as pd

# Customers
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5],
    "customer_name": ["John", "Mary", "David", "Alice", "Bob"],
    "city": ["Mumbai", "Delhi", "Pune", "Mumbai", "Chennai"]
})

# Products
products = pd.DataFrame({
    "product_id": [101, 102, 103, 104, 105],
    "product_name": ["Laptop", "Phone", "Tablet", "Headphones", "Keyboard"],
    "category": ["Electronics", "Electronics", "Electronics", "Accessories", "Accessories"],
    "price": [60000, 30000, 20000, 3000, 1500]
})

# Orders
orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005],
    "customer_id": [1, 2, 3, 4, 5],
    "product_id": [101, 102, 103, 104, 105],
    "quantity": [1, 2, 1, 3, 2]
})

# Payments
payments = pd.DataFrame({
    "payment_id": [501, 502, 503, 504, 505],
    "order_id": [1001, 1002, 1003, 1004, 1005],
    "payment_status": ["Paid", "Paid", "Paid", "Pending", "Paid"]
})

print("Customers:")
display(customers)

print("Products:")
display(products)

print("Orders:")
display(orders)

print("Payments:")
display(payments)
```

```python
customers.to_csv("customers.csv", index=False)
products.to_csv("products.csv", index=False)
orders.to_csv("orders.csv", index=False)
payments.to_csv("payments.csv", index=False)

print("All CSV files created successfully!")
```

Data Cleaning.

```python
# Check for missing values

print("Missing values in Customers:")
print(customers.isnull().sum())

print("\nMissing values in Products:")
print(products.isnull().sum())

print("\nMissing values in Orders:")
print(orders.isnull().sum())

print("\nMissing values in Payments:")
print(payments.isnull().sum())
```

```python
# Check for missing values

print("Missing values in Customers:")
print(customers.isnull().sum())

print("\nMissing values in Products:")
print(products.isnull().sum())

print("\nMissing values in Orders:")
print(orders.isnull().sum())

print("\nMissing values in Payments:")
print(payments.isnull().sum())
```

```python
# Remove duplicate records

customers = customers.drop_duplicates()
products = products.drop_duplicates()
orders = orders.drop_duplicates()
payments = payments.drop_duplicates()

print("Duplicates removed successfully!")
```

```python
print("Customers data types:")
print(customers.dtypes)

print("\nProducts data types:")
print(products.dtypes)

print("\nOrders data types:")
print(orders.dtypes)

print("\nPayments data types:")
print(payments.dtypes)
```

```python
print("Cleaned Customers:")
display(customers)

print("Cleaned Products:")
display(products)

print("Cleaned Orders:")
display(orders)

print("Cleaned Payments:")
display(payments)
```

Data Transformation.

```python
order_data = orders.merge(
    products,
    on="product_id",
    how="inner"
)

print("Orders + Products:")
display(order_data)
```

```python
order_data["total_sales"] = (
    order_data["price"] * order_data["quantity"]
)

print("Total Sales Calculated:")
display(order_data)
```

```python
order_data = order_data.merge(
    customers,
    on="customer_id",
    how="inner"
)

print("Added Customer Information:")
display(order_data)
```

```python
order_data = order_data.merge(
    payments,
    on="order_id",
    how="inner"
)

print("Final Transformed Dataset:")
display(order_data)
```

```python
sales_summary = order_data.groupby("category").agg(
    total_sales=("total_sales", "sum"),
    total_quantity=("quantity", "sum")
).reset_index()

print("Sales Summary by Category:")
display(sales_summary)
```

Data Storage in SQL Database.

```python
import sqlite3

# Create SQLite database
connection = sqlite3.connect("ecommerce_warehouse.db")

print("Database created successfully!")
```

```python
customers.to_sql(
    "dim_customer",
    connection,
    if_exists="replace",
    index=False
)

products.to_sql(
    "dim_product",
    connection,
    if_exists="replace",
    index=False
)

orders.to_sql(
    "orders",
    connection,
    if_exists="replace",
    index=False
)

payments.to_sql(
    "payments",
    connection,
    if_exists="replace",
    index=False
)

order_data.to_sql(
    "fact_sales",
    connection,
    if_exists="replace",
    index=False
)

print("All tables stored in the database!")
```

```python
print("Customer Dimension:")
display(pd.read_sql_query(
    "SELECT * FROM dim_customer",
    connection
))

print("Product Dimension:")
display(pd.read_sql_query(
    "SELECT * FROM dim_product",
    connection
))

print("Fact Sales:")
display(pd.read_sql_query(
    "SELECT * FROM fact_sales",
    connection
))
```

```python
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table'",
    connection
)

print("Tables in Data Warehouse:")
display(tables)
```

Reporting

```python
query = """
SELECT SUM(total_sales) AS total_sales
FROM fact_sales
"""

total_sales_report = pd.read_sql_query(
    query,
    connection
)

print("Total Sales:")
display(total_sales_report)
```

```python
query = """
SELECT category,
       SUM(total_sales) AS total_sales
FROM fact_sales
GROUP BY category
"""

category_report = pd.read_sql_query(
    query,
    connection
)

print("Sales by Category:")
display(category_report)
```

```python
query = """
SELECT customer_name,
       SUM(total_sales) AS total_sales
FROM fact_sales
GROUP BY customer_name
"""

customer_report = pd.read_sql_query(
    query,
    connection
)

print("Sales by Customer:")
display(customer_report)
```

```python
query = """
SELECT payment_status,
       COUNT(*) AS number_of_orders,
       SUM(total_sales) AS total_sales
FROM fact_sales
GROUP BY payment_status
"""

payment_report = pd.read_sql_query(
    query,
    connection
)

print("Payment Report:")
display(payment_report)
```

Chart

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    category_report["category"],
    category_report["total_sales"]
)

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```
